# Z test

Suppose you have a a sample $N$ of people. You want to infer some information on the average height of the *General* population mean.

```python
N = 50 # sample size
mu = np.random.randint(170,180)
sigma = 7.5
L = np.random.normal(mu, sigma,N)
```

Supposing a normal distribution as a good approximation when we have many data points from the Central Limit theorem. 

But it's good to recall it :  Let $X_1, X_2 , \dots , X_N$ **independent & identically distributed random variables** with EV $\mu$ and finite variance $\sigma^2$ $$\frac{X_1 + \dots + X_N}{n} \rightarrow \mu  \text{ a.s} $$

```python
## our hypothesis mean height for this sample, ie H0 : mu = mu0
mu0 = 176
```

```python
z = np.sqrt(N)*(np.mean(L) - mu0)/ sigma
print(z)
```

We choose a confidence level $\alpha$, think of it as an error risk budget. ie you can afford to be wrong for example $5\%$ of the time or $1\%$ of the time. Intuitively, If you need to be wrong less than 1% of the time, you need extraordinary evidence, but being wrong less than 5% of the time is less conservative which relaxes the texting.

In our case, the test statistic we get of $-5.869$ is very large, in fact its about 6 deviations lower than mean, a **6-sigma** event is very rare in real life (odds 900,000,000 : 1), so we are more likely to be wrong about our hypothesis than reality to produce a 6-sigma event. We reject the hypothesis. The true randomly generated mean is not $176$. The resulting z-score is negative so we are likely to be overestimating the true mean. we update on the downside.

```python
mu0 = 174
z = np.sqrt(N)*(np.mean(L) - mu0)/ sigma
print(z)
```

Again, the test statistic we get of $-3.824$ is very large, in fact its about 3 deviations lower than mean, a **3-sigma** event is also rare (odds 370 : 1), so we are more likely to be wrong about our hypothesis. We reject the hypothesis. The true randomly generated mean is not $174$. The resulting z-score is negative so we are likely to be overestimating the true mean. we update on the downside.

```python
mu0 = 170
z = np.sqrt(N)*(np.mean(L) - mu0)/ sigma
print(z)
```

The test statistic we get of $0$ is approximately null, this is not very far from the mean, and very likely, so we are more likely to be right about our hypothesis. We don't reject the hypothesis. The true randomly generated mean is not necessarily $170$. But Assuming it is will not hurt us. let's Check :  

```python
print(mu)
```

# Two-Sample test : Two classes, One professor. 
In Evaluating the performance of the classes, we can take the grades of the classes and see how they differ, as the professor is providing the same teaching, any difference is likely to be from the ability of students.

Suppose now you have two classes with different distributions of grades. You want to determine whether the classes have the same mean.

```python
N = 30 # same number of students
mu1 = np.random.uniform(16,17)
mu2 = np.random.uniform(16,17)
sigma = 2.46 # equal variance case


C1 = np.random.normal(mu1, sigma ,N)
C2 = np.random.normal(mu2, sigma ,N)

t = (np.mean(C1) - np.mean(C2)) / (np.sqrt(1/N) * sigma)
print(t)
```

```python
quantile = stats.t.ppf([0.025,0.975], df= N-2) # 5% confidence level
print(quantile) 
```

```python
mu_s1 = np.mean(C1)
mu_s2 = np.mean(C2)

print(f"Mean of C1 : {mu_s1:.2f}")
print(f"Mean of C2 : {mu_s2:.2f}")
```

Our Test gives $-2.455 < -2.04$ we can reject the equality hypothesis, the classes are not attaining the same grades even with the same teaching, there is something wrong.

### Rejection : A systematic Rigorous approach
Previously we kept updating $\mu_0$ until we found a value where we don't reject. Now we will take a look at a concept called the significance level $\alpha$ the confidence level $1-\alpha$, Power $1-\beta$ and p-value.

- $\alpha$ is the probability of rejecting the Null when it is actually True. (Type I error)
- $\beta$ is probability of failing to reject the Null when it's False. (Type II error)

To not get confused, think of yourself as a engineer designing a cancer detection system, a Structural Engineer designing a bridge. What do you think you should Assume the status quo ?
In System design we like robustness, meaning we don't want to assert no presence of cancer where it actually is, and we don't design bridges thinking they won't fail, but thinking they will.

**H0** would be then an equality statement : Cancer **is** present - Bridge **will** fall , and we have to find Evidence otherwise.

- In the case of the bridge for example, Rejecting "Bridge will Fall" when it's actually true is fatal, it will kill god forbid, hundreds of people. So we want to minimize the probability of making this error. ie We want to minimize The probability of rejecting while it's actually True. The significant level would be something like 0.01. Meaning that you need extraordinary evidence of the strength of the bridge to build it. In the case of the patient, failing to reject the null when it's false, ie telling him he's sick while he isn't is scary, but not fatal you can always do a second look.

This is not a one way optimization and there is a general tradeoff to be made. As type I and type II error pull our error budget in two opposing ways.
The more we want to minimize type 1 error, we reject less, meaning we increase the probability of failing to reject, and the more we want to minimize type 2 error, we reject more, increasing probability of rejecting when it's actually True.

### Unequal Variance case : 
It geniunely gets harder to establish a results as the uncertainty from the unequal variances, the estimations make the confidence intervals more expanded and harder to reject.

```python
N = 30 # same number of students
mu1 = np.random.uniform(16,17)
mu2 = np.random.uniform(16,17)
sigma1 = np.random.uniform(2,3) # equal variance case
sigma2 = np.random.uniform(2,3)


C1 = np.random.normal(mu1, sigma1 ,N)
C2 = np.random.normal(mu2, sigma2 ,N)

t = (np.mean(C1) - np.mean(C2)) / np.sqrt((np.std(C1)**2 +np.std(C1)**2)/ N)
print(t)
```

```python
quantile = stats.t.ppf([0.025,0.975], df= N-2) # 5% confidence level
print(quantile) 
```

```python
mu_s1 = np.mean(C1)
mu_s2 = np.mean(C2)

print(f"Mean of C1 : {mu_s1:.2f}")
print(f"Mean of C2 : {mu_s2:.2f}")
```

Our Test gives $-3.02 < -2.04$ we can reject the equality hypothesis, the classes are not attaining the same grades even with the same teaching, there is something wrong.
